```python
import pandas as pd
import numpy as np

# Load raw dataset
df = pd.read_excel("../data/raw/indian-job-market-dataset-2025.xlsx")

# Create working copy
df_clean = df.copy()

# Remove exact duplicate records
df_clean = df_clean.drop_duplicates().copy()

# Handle missing company names
df_clean["companyName"] = df_clean["companyName"].fillna("Unknown Company")

# Create standardized experience field
df_clean["experience_clean"] = df_clean["experience"]

mask = (
    df_clean["experience_clean"].isna()
    & df_clean["minimumExperience"].notna()
    & df_clean["maximumExperience"].notna()
)

df_clean.loc[mask, "experience_clean"] = (
    df_clean.loc[mask, "minimumExperience"].astype(int).astype(str)
    + "-"
    + df_clean.loc[mask, "maximumExperience"].astype(int).astype(str)
    + " Yrs"
)

df_clean["experience_clean"] = df_clean["experience_clean"].fillna(
    "Not Specified"
)

print("Dataset ready:", df_clean.shape)
```

```python
df_clean["tagsAndSkills"].head(20)
```

```python
def clean_skills(value):
    if pd.isna(value):
        return []
    
    skills = str(value).split(",")
    
    skills = [
        skill.strip().lower()
        for skill in skills
        if skill.strip()
    ]
    
    return list(dict.fromkeys(skills))
```

```python
df_clean["skills_clean"] = df_clean["tagsAndSkills"].apply(clean_skills)
```

```python
# cateogorising the roles
def classify_role(title):
    title = str(title).lower()

    if any(x in title for x in [
        "data engineer", "data engineering", "etl",
        "big data", "data warehouse", "data platform"
    ]):
        return "Data Engineering"

    elif any(x in title for x in [
        "data analyst", "business analyst", "bi analyst",
        "business intelligence", "reporting analyst"
    ]):
        return "Data Analytics"

    elif any(x in title for x in [
        "machine learning", "machine learning engineer",
        "ai engineer", "artificial intelligence",
        "ml engineer", "deep learning", "nlp"
    ]):
        return "AI / ML"

    elif any(x in title for x in [
        "software engineer", "software developer",
        "application developer", "java developer",
        "python developer", "full stack", "backend",
        "frontend", "web developer"
    ]):
        return "Software Development"

    elif any(x in title for x in [
        "devops", "cloud engineer", "cloud",
        "site reliability", "sre", "infrastructure"
    ]):
        return "Cloud / DevOps"

    else:
        return "Other"
```

```python
df_clean["role_category"] = df_clean["title"].apply(classify_role)
```

```python
df_clean["role_category"].value_counts()
```

```python
candidate_skills = [
    "python",
    "sql",
    "pandas",
    "power bi"
]

candidate_role = "Data Engineering"
candidate_location = "Hyderabad"
candidate_experience = 0
```

```python
candidate_jobs = df_clean[
    (df_clean["role_category"] == candidate_role) &
    (df_clean["minimumExperience"] <= candidate_experience + 1)
].copy()
```

```python
candidate_jobs.shape
```

```python
def calculate_skill_match(job_skills, candidate_skills):
    if not job_skills:
        return 0
    
    matched = set(job_skills) & set(candidate_skills)
    return round((len(matched) / len(set(job_skills))) * 100, 2)
```

```python
candidate_jobs["skill_match"] = candidate_jobs["skills_clean"].apply(
    lambda x: calculate_skill_match(x, candidate_skills)
)
```

```python
candidate_jobs = candidate_jobs.sort_values(
    "skill_match",
    ascending=False
)
```

```python
candidate_jobs[
    ["title", "companyName", "location", "skills_clean", "skill_match"]
].head(10)
```

```python
def location_score(job_location, candidate_location):
    job_location = str(job_location).lower()
    candidate_location = candidate_location.lower()

    if candidate_location in job_location:
        return 100
    elif "remote" in job_location:
        return 80
    else:
        return 0
```

```python
candidate_jobs["location_score"] = candidate_jobs["location"].apply(
    lambda x: location_score(x, candidate_location)
)
```

```python
def experience_score(min_exp, candidate_exp):
    if pd.isna(min_exp):
        return 50
    
    if min_exp <= candidate_exp:
        return 100
    elif min_exp == candidate_exp + 1:
        return 80
    elif min_exp == candidate_exp + 2:
        return 50
    else:
        return 20
```

```python
candidate_jobs["experience_score"] = candidate_jobs[
    "minimumExperience"
].apply(
    lambda x: experience_score(x, candidate_experience)
)
```

```python
candidate_jobs["final_score"] = (
    candidate_jobs["skill_match"] * 0.60
    + candidate_jobs["location_score"] * 0.25
    + candidate_jobs["experience_score"] * 0.15
)
```

```python
candidate_jobs = candidate_jobs.sort_values(
    "final_score",
    ascending=False
)
```

```python
candidate_jobs = candidate_jobs[
    candidate_jobs["skill_match"] > 0
].copy()
```

```python
candidate_jobs["final_score"] = (
    candidate_jobs["skill_match"] * 0.70
    + candidate_jobs["experience_score"] * 0.20
    + candidate_jobs["location_score"] * 0.10
)
```

```python
candidate_jobs = candidate_jobs.sort_values(
    "final_score",
    ascending=False
)
```

```python
candidate_jobs[
    [
        "title",
        "companyName",
        "location",
        "skill_match",
        "location_score",
        "experience_score",
        "final_score"
    ]
].head(10)
```

```python
def get_skill_gaps(job_skills, candidate_skills):
    job_set = set(job_skills)
    candidate_set = set(candidate_skills)

    return sorted(job_set - candidate_set)
```

```python
candidate_jobs["skill_gaps"] = candidate_jobs["skills_clean"].apply(
    lambda x: get_skill_gaps(x, candidate_skills)
)
```
